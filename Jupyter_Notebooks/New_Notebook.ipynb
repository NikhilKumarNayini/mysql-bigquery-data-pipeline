```python
import requests
url = "https://datahub.io/core/airport-codes/r/airport-codes.csv"
local_path = "/tmp/airport_codes.csv"


def download_file(url: str, local_path: str):
    """Downloads any file (CSV/JSON/etc.) to local path."""
    data = requests.get(url)
    data.raise_for_status()
    with open(local_path, "wb") as f:
        f.write(data.content)
    print(f"Downloaded → {local_path}")

download_file(url, local_path)
```

```python
def load_to_spark(path: str, fmt: str = None):

    """    Load a file into Spark based on extension or provided format.Supports CSV, JSON, Parquet."""
    if fmt is None:
        fmt = path.split(".")[-1]  # detect by extension

    if fmt.lower() == "csv":
        return spark.read.csv(path, header=True, inferSchema=True)
    if fmt.lower() == "json":
        return spark.read.json(path)

    if fmt.lower() == "parquet":
        return spark.read.parquet(path)

    raise ValueError(f"Unsupported format: {fmt}")

df_spark = load_to_spark(local_path)
df_spark.show(20, False)
```

```python
data_employee = [
    (101, "Rajesh", "IT", 75000, "Bangalore"),
    (102, "Priya", "HR", 60000, "Mumbai"),
    (103, "Anil", "IT", 80000, "Hyderabad"),
    (104, "Sneha", "HR", 62000, "Pune"),
    (105, "Manish", "Finance", 90000, "Chennai"),
    (106, "Suresh", "IT", 78000, "Bangalore")
]

employee_columns = ["emp_id", "name", "department", "salary", "city"]

df = spark.createDataFrame(data_employee, employee_columns)
df.show()
```

```python
# Example transformation: Calculate bonus (10% of salary for people who have salary upto 75000, 15% salary for people who have salary more than 75000)
df_transformed = df.withColumn(
    "bonus",
    when(col("salary") < 75000, col("salary") * 0.1)
    .otherwise(col("salary") * 0.15)
)


# Filter employees with salary greater than 70,000
df_transformed = df_transformed.withColumn("AggregatedSalary", col("salary") + col("bonus"))
df_transformed.show(10, False)
df_transformed = df_transformed.filter(df["salary"] > 70000)
```
